```python

import pandas as pd
import numpy as np

pasta = '/content/drive/MyDrive/tblh agraria minoro/'

real = pd.read_excel(pasta + 'Realizado.xlsx')
orc = pd.read_excel(pasta + 'Orcado.xlsx', sheet_name='Orçado 26')

# Conversão de datas
real['Dta.liq.'] = pd.to_datetime(real['Dta.liq.'], errors='coerce')

print('=== QUALIDADE DOS DADOS ===')
print('Registros:', len(real))
print('Fretes únicos:', real['Nº frete'].nunique())
print('Datas:', real['Dta.liq.'].min(), 'até', real['Dta.liq.'].max())
print('Valores ausentes:')
print(real[['Nº frete', 'Valor líquido',
            'Peso líquido', 'Valor líquido.1']].isna().sum())

print('\n=== REPETIÇÃO POR FRETE ===')
repeticoes = real.groupby('Nº frete').size()
print(repeticoes.describe())
print('Fretes com mais de uma linha:', (repeticoes > 1).sum())

print('\n=== VALORES POR FRETE ===')
analise = real.groupby('Nº frete').agg(
    linhas=('Item', 'size'),
    valor_liquido=('Valor líquido', 'nunique'),
    valor_liquido_1=('Valor líquido.1', 'nunique'),
    peso=('Peso líquido', 'sum'),
    peso_acumulado=('Peso líq.acumul.', 'nunique')
)

display(analise.head(10))

print('\n=== RESUMO DAS COLUNAS FINANCEIRAS ===')
display(real[['Valor líquido', 'Valor líquido.1',
              'Montante do imposto', 'Peso líquido']].describe())

print('\n=== ORÇAMENTO ===')
print('Registros:', len(orc))
display(orc[['mês', 'FRETE', 'VOLUME', 'R$/ton']].describe())
```
